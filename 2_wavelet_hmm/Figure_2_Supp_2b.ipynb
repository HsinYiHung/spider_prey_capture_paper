```python
import numpy as np
from scipy.interpolate import interp1d

def compute_normalized_time_accuracy_weighted(y_true, y_pred, states=[0,1], n_points=100):
    """
    Compute length-weighted accuracy curves over normalized time for each state.

    Parameters:
        y_true (np.array): True labels
        y_pred (np.array): Predicted labels
        states (list): list of states to analyze
        n_points (int): number of normalized time points

    Returns:
        mean_acc (dict): state -> weighted mean accuracy curve (n_points,)
        norm_times (np.array): normalized time axis
        raw_acc (dict): state -> raw accuracy (single number, unnormalized)
    """
    def find_episodes(state_array, target_state):
        episodes = []
        in_episode = False
        start_idx = 0
        for i, s in enumerate(state_array):
            if s == target_state:
                if not in_episode:
                    in_episode = True
                    start_idx = i
            else:
                if in_episode:
                    episodes.append((start_idx, i-1))
                    in_episode = False
        if in_episode:
            episodes.append((start_idx, len(state_array)-1))
        return episodes

    acc_curves = {s: [] for s in states}
    weights = {s: [] for s in states}

    for state in states:
        episodes = find_episodes(y_true, state)
        for start, end in episodes:
            length = end - start + 1
            if length < 2:
                continue

            # accuracy vector for this episode
            acc_episode = (y_pred[start:end+1] == y_true[start:end+1]).astype(float)
            orig_times = np.linspace(0, 1, length)

            # interpolate to normalized time
            interp_fn = interp1d(orig_times, acc_episode, kind='nearest', 
                                 bounds_error=False, fill_value='extrapolate')
            norm_times = np.linspace(0, 1, n_points)
            acc_interp = interp_fn(norm_times)

            acc_curves[state].append(acc_interp)
            weights[state].append(length)  # weight = episode length

    # length-weighted average accuracy curve
    mean_acc = {}
    raw_acc = {}
    for state in states:
        if len(acc_curves[state]) == 0:
            mean_acc[state] = np.full(n_points, np.nan)
            raw_acc[state] = np.nan
            continue

        acc_array = np.array(acc_curves[state])   # shape (n_episodes, n_points)
        w_array = np.array(weights[state])[:, None]  # shape (n_episodes, 1)

        weighted_sum = np.nansum(acc_array * w_array, axis=0)
        total_weight = np.sum(w_array)
        mean_acc[state] = weighted_sum / total_weight

        # raw accuracy without normalization
        correct_total = np.sum((y_pred[y_true == state] == state).astype(float))
        total_state = np.sum(y_true == state)
        raw_acc[state] = correct_total / total_state

    return mean_acc, np.linspace(0, 1, n_points), raw_acc
```
